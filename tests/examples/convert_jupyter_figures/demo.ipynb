# Figures Demo

The figures in the outputs are written to files next to the converted notebook.

```python
import matplotlib.pyplot as plt

with open("count.txt", encoding="utf-8") as fh:
    count = int(fh.read())
```

```python
for i in range(count):
    fig, ax = plt.subplots(figsize=(2, 2))
    ax.plot([0, 1], [i, 1 - i])
    plt.show()
```
