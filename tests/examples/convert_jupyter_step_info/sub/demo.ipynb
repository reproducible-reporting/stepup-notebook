```python
import helper

from stepup.core.api import amend
```

```python
amend(inp="data.txt", out="result.txt")
with open("data.txt", encoding="utf-8") as fh:
    value = int(fh.read())
with open("result.txt", "w", encoding="utf-8", newline="\n") as fh:
    fh.write(f"{helper.double(value)}\n")
```
